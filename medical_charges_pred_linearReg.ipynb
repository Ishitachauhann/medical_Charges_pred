```python
import requests
import certifi

# URL for the medical charges dataset
medical_charges_url = 'https://raw.githubusercontent.com/JovianML/opendatasets/master/data/medical-charges.csv'

# Send the request and verify using certifi's certificates
response = requests.get(medical_charges_url, verify=certifi.where())

# Check if the request was successful
if response.status_code == 200:
    # Save the content to a local file
    with open('medical.csv', 'wb') as file:
        file.write(response.content)
    print("Download complete: 'medical.csv' saved successfully.")
else:
    print(f"Failed to download the file. Status code: {response.status_code}")
```

eqn of line = y=wx+c
here , 
charges = w(age)+ b 

thsi si a linear regression model 
- the numbers w and b are the parameters or weights of teh model 
- values of age ---> inputs of the model and 
- values of charges ---> targets 

defining a function to compute charges givena ge ,w,b 

```python
def estimate_charges(age,w,b):
    return w*age + b 
```

teh estimate_charges func is our veyr first model ,,,

```python
def rmse(targets , predictions):
    return np.sqrt(np.mean(np.square(targets-predictions)))
```

#### modify try_parameter

```python
def try_parameters(w, b):
    ages = non_smoker_df.age
    target = non_smoker_df.charges

    predictions = estimate_charges(ages, w, b)

    plt.plot(ages, predictions, 'r', alpha=0.9)
    plt.scatter(ages, target, s=8, alpha=0.9)
    plt.xlabel('age')
    plt.ylabel('charges')
    plt.legend(['prediction', 'actual'])
    plt.show()  # <-- Optional, but useful in some environments
    
    loss = rmse(target,predictions)
    print("rmse loss :",loss)
```

```python
try_parameters(50,100)
```
